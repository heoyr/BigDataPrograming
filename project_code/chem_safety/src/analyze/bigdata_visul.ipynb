```python
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
import numpy as np

plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

labels = ['표백제', '접착제', '세정제', '합성세제',
          '방충제', '방향제', '탈취제', '코팅제']
counts = [211, 126, 99, 91, 35, 32, 29, 16]
colors = ['#B45454', '#C47A45', '#C47A45', '#A07830',
          '#2C5FAD', '#2C5FAD', '#2C5FAD', '#64748B']

fig, ax = plt.subplots(figsize=(10, 6))
fig.patch.set_facecolor('white')
ax.set_facecolor('white')

y_pos = np.arange(len(labels))
bars = ax.barh(y_pos, counts, color=colors,
               height=0.6, edgecolor='white', linewidth=2)

ax.set_yticks(y_pos)
ax.set_yticklabels(labels, fontsize=13)
ax.set_title('품목별 화학사고 발생 건수 (총 678건)',
             fontsize=14, fontweight='bold', color='#1B2B5E', pad=15)
ax.set_xlim(0, 260)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_visible(False)
ax.spines['bottom'].set_visible(False)
ax.tick_params(left=False, bottom=False)
ax.set_xticks([])
ax.set_facecolor('#F8FAFC')

for i in range(len(labels)):
    ax.axhspan(i - 0.38, i + 0.38,
               color='white', alpha=1, zorder=0)
    ax.axhline(y=i - 0.4, color='#E2E8F0',
               linewidth=0.8, zorder=0)

for bar, cnt, color in zip(bars, counts, colors):
    ax.text(bar.get_width() + 5,
            bar.get_y() + bar.get_height() / 2,
            str(cnt) + '건',
            va='center', ha='left',
            fontsize=12, fontweight='bold',
            color=color)

plt.tight_layout()
plt.savefig('q1_accident.png', dpi=150,
            bbox_inches='tight', facecolor='white')
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

df = pd.DataFrame({
    '구분': ['전성분 공개 기업', '미공개 기업'],
    '위반율': [0.8, 46.7],
    '전체기업수': [129, 8615],
    '위반기업수': [1, 4026]
})

fig, ax = plt.subplots(figsize=(7, 5.5))
fig.patch.set_facecolor('white')
ax.set_facecolor('white')

colors = ['#3D7A5A', '#B45454']
x = np.arange(len(df))
bars = ax.bar(x, df['위반율'], color=colors, width=0.45,
              edgecolor='white', linewidth=1.5, zorder=3)

ax.set_ylabel('위반율 (%)', fontsize=12, color='#64748B')
ax.set_title('Q2. 전성분 공개 기업 vs 미공개 기업 위반율',
             fontsize=13, fontweight='bold', color='#1B2B5E', pad=15)
ax.set_xticks(x)
ax.set_xticklabels(
    [f'전성분 공개 기업\n(총 {df.전체기업수[0]:,}개사)',
     f'미공개 기업\n(총 {df.전체기업수[1]:,}개사)'],
    fontsize=11)
ax.set_ylim(0, 58)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.grid(axis='y', color='#E2E8F0', linewidth=0.8, zorder=0)

for bar, rate, color in zip(bars, df['위반율'], colors):
    ax.text(bar.get_x() + bar.get_width() / 2,
            bar.get_height() + 0.8,
            str(rate) + '%',
            ha='center', fontsize=16,
            fontweight='bold', color=color)

for bar, row in zip(bars, df.itertuples()):
    ax.text(bar.get_x() + bar.get_width() / 2,
            bar.get_height() / 2,
            f'위반 {row.위반기업수:,}개사',
            ha='center', va='center',
            fontsize=10, color='#FFD700', fontweight='bold')

plt.tight_layout()
plt.savefig('q2_violation_rate.png', dpi=150,
            bbox_inches='tight', facecolor='white')
plt.show()
```
